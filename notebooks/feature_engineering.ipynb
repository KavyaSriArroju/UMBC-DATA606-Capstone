```python
import pandas as pd
import numpy as np

# Load data
df = pd.read_csv("PJME_hourly.csv")

# Convert datetime
df["Datetime"] = pd.to_datetime(df["Datetime"])
df = df.sort_values("Datetime").set_index("Datetime")

# -------------------
# TIME FEATURES
# -------------------
df["hour"] = df.index.hour
df["dayofweek"] = df.index.dayofweek
df["month"] = df.index.month
df["year"] = df.index.year
df["is_weekend"] = df["dayofweek"].isin([5, 6]).astype(int)

# -------------------
# CYCLICAL FEATURES (advanced)
# -------------------
df["hour_sin"] = np.sin(2 * np.pi * df["hour"] / 24)
df["hour_cos"] = np.cos(2 * np.pi * df["hour"] / 24)

df["month_sin"] = np.sin(2 * np.pi * df["month"] / 12)
df["month_cos"] = np.cos(2 * np.pi * df["month"] / 12)

# -------------------
# LAG FEATURES (VERY IMPORTANT)
# -------------------
df["lag_1"] = df["PJME_MW"].shift(1)
df["lag_24"] = df["PJME_MW"].shift(24)
df["lag_168"] = df["PJME_MW"].shift(168)

# -------------------
# ROLLING FEATURES
# -------------------
df["rolling_mean_24"] = df["PJME_MW"].rolling(24).mean()
df["rolling_std_24"] = df["PJME_MW"].rolling(24).std()

# -------------------
# DROP NA
# -------------------
df = df.dropna()

# -------------------
# SAVE FILE
# -------------------
df.to_csv("featured_data.csv")

df.head()
```

### Feature Engineering

Time-based features such as hour, day of week, and month were created to capture temporal patterns.  
Lag features (1 hour, 24 hours, and 168 hours) were added to model past dependencies.  
Rolling statistics were used to capture short-term trends.  
Cyclical encoding was applied to represent periodic time patterns.
